## **Practical 4** : Perform Exploratory Data Analysis (EDA) using descriptive statistics and identify trends in datasets

Step 1 import required libraries

```python
# Import Pandas library
import pandas as pd

# Import Matplotlib library
import matplotlib.pyplot as plt
```

Step 2 Create Dataset

```python
# Create a dictionary containing student data
data = {
    "Month" : ["Januray" , "February" , "March" , "April" , "May" , "June" , "July" , "August" , "September" , "October" , "November" , "December"],
    "Units_Sold" : [120,135,150,145,160,175,180,170,190,210,230,250],
    "Revenue" : [120000,135000,150000,145000,160000,175000,180000,170000,190000,210000,230000,250000],
    "Profit" : [20000,23000,25000,24000,27000,30000,31000,29000,33000,37000,41000,45000]
}
```

```python
# Create a DataFrame from the dictionary 'data'
df = pd.DataFrame(data)
# Display the DataFrame
df
```

Step 3 Inspect Dataset

```python
#Display the first five records
df.head()
```

```python
# Display the number of rows and columns in the DataFrame
df.shape
```

```python
# Display all column names in the DataFrame
df.columns
```

```python
# Display information about the DataFrame
df.info()
```

Step 4 Generate Descriptive Statistic

```python
# Display statistical summary of numerical data
df.describe()
```

Step 5 Calculate Mean

```python

# Calculate the average value of the Units_Sold column
average_units = df["Units_Sold"].mean()

# Display the average Units Sold
print("Average Units Sold:", average_units)
```

Step 6 Calculate Median

```python
# Calculate the median value of the Units_Sold column
median_units = df["Units_Sold"].median()

# Display the median Units Sold
print("Median Units Sold:", median_units)
```

Step 7 Calculate Maximum and Minmum

```python
# Find the maximum value in the Units_Sold column
maximum_units = df["Units_Sold"].max()

# Find the minimum value in the Units_Sold column
minimum_units = df["Units_Sold"].min()

# Display the maximum Units Sold
print("Maximum Units Sold:", maximum_units)

# Display the minimum Units Sold
print("Minimum Units Sold:", minimum_units)
```

Step 8 Calculate Standard Deviation

```python
#Calculate the standard devi
std_units = df["Units_Sold"].std()

# Display the Result
print("Standard Deviation of Units Sold:", std_units)
```

Step 9 Find the Month with highest sale

```python
# Find the index of the row having the maximum revenue.
highest_sales_index = df["Revenue"].idxmax()

#Retrive the month corresponding to that index.
highest_sales_month = df.loc[highest_sales_index, "Month"]

#Retrive the maximum revenue.
highest_revenue = df.loc[highest_sales_index, "Revenue"]

#Display the Result
print("Highest Revenue:", highest_revenue)
print("Highest Sales Month:", highest_sales_month)
```

Step 10 Find the month with lowest sale

```python
# Find the index of the row having the maximum revenue.
lowest_sales_index = df["Revenue"].idxmin()

#Retrive the month corresponding to that index.
lowest_sales_month = df.loc[lowest_sales_index, "Month"]

#Retrive the maximum revenue.
lowest_revenue = df.loc[lowest_sales_index, "Revenue"]

#Display the Result
print("Lowest Revenue:", lowest_revenue)
print("Lowest Sales Month:", lowest_sales_month)
```

Step 11 Calculate Total Revenues

```python
# calculate the total revenue generated during the year
total_revenue = df["Revenue"].sum()

# Display the Result
print("Total Revenue:", total_revenue)
```

Step 12 Calculate total profit

```python
# calculate the total profit generated during the year
total_profit = df["Profit"].sum()

# Display the Result
print("Total Profit:", total_profit)
```

Step 13 Identify the sales trend

```python
from matplotlib import lines
# Create a line chart to Visualize monthly revenue
plt.figure(figsize=(10, 5))

# plot month on the x-axis and revenue on the y-axis
plt.plot(df["Month"], df["Revenue"], marker='o' , linestyle=':')

# Add a title to the graph

#Add a title to graph
plt.title("Monthly Revenue Trend")

# add title in x-axis
plt.xlabel("Month")

# add title in y-axis
plt.ylabel("Revenue")

# rotate
plt.xticks(rotation=45)

# Display the graph
plt.show()

```

Step  14 Analye Profit

```python
# Create a line chart to Visualize monthly revenue
plt.figure(figsize=(10, 5))

# plot month on the x-axis and profit on the y-axis
plt.plot(df["Month"], df["Profit"], marker='o' )

#Add a title to graph
plt.title("Monthly Profit Trend")

# add label in x-axis
plt.xlabel("Month")

# add label in y-axis
plt.ylabel("Profit")

# rotate
plt.xticks(rotation=45)

# Display the graph
plt.show()
```

Step 15 Compare Revenue and Profit

```python
plt.figure(figsize=(10, 5))

# Plot Revenue
plt.plot(df["Month"], df["Revenue"], marker='o', label="Revenue")

# Plot Profit
plt.plot(df["Month"], df["Profit"], marker='o', label="Profit")

# Add a title to the graph
plt.title("Compare Revenue and Profit")

# Add labels to the x-axis and y-axis
plt.xlabel("Month")
plt.ylabel("Amount")

# Add a legend to the graph
plt.legend()

#rotate monthly
plt.xticks(rotation=45)

# Display the graph
plt.show()

```

Step 16 identify growth in sales

```python
# Calculate the percentage change in revenue
#from one month to the next.
df["Revenue_Growth_%"] = df["Revenue"].pct_change() * 100

# Display the updated dataset
df
```

Step 17 Identify months with increasing revenue

```python
# Filter the dataset to display only those months
# where revenue growth is positive
increasing_Sales = df[df["Revenue_Growth_%"] > 0]

# Display the months with increasing revenue
increasing_Sales
```
