**Step 1**: Import Libraries

```python
# 1. Import Libraries
import pandas as pd
import numpy as np
import plotly.express as px

# Install missing Dash package
!pip install dash -q
from dash import Dash, dcc, html, dash_table
from dash.dependencies import Input, Output
```

```python
df = pd.read_csv("Coffe_sales.csv")
df
```

**Step 3**: Cleaning The dataset

```python
# Convert Date into dd-mm-yyyy format
df["Date"] = pd.to_datetime(
    df["Date"], format="mixed", dayfirst=True
).dt.strftime("%d-%m-%Y")

# Convert Time into HH:MM:SS format
df["Time"] = pd.to_datetime(
    df["Time"], format="mixed"
).dt.strftime("%H:%M:%S")

# Display cleaned dataset
display(df)

# Check data types
print(df.dtypes)
```

**Step 5**: Create Dashboard layout

```python
#  Initialize Dash App
app = Dash(__name__)
app.title = "Coffee Sales Dashboard"

# Dropdown Options
coffee_options = [
    {"label": "All Coffee", "value": "All"}
] + [
    {"label": coffee, "value": coffee}
    for coffee in sorted(df["coffee_name"].unique())
]

time_options = [
    {"label": "All Time Periods", "value": "All"}
] + [
    {"label": period, "value": period}
    for period in sorted(df["Time_of_Day"].unique())
]

cash_options = [
    {"label": "All Payment Methods", "value": "All"}
] + [
    {"label": cash.capitalize(), "value": cash}
    for cash in sorted(df["cash_type"].unique())
]

# 6. Dashboard Layout
app.layout = html.Div(
    style={
        "backgroundColor": "#f5f3ef",
        "padding": "25px",
        "fontFamily": "Arial"
    },

    children=[

        # Dashboard Heading
        html.Div([
            html.H1(
                "☕ COFFEE SALES DASHBOARD",
                style={
                    "textAlign": "center",
                    "color": "#4b2815",
                    "fontSize": "32px"
                }
            ),

            html.P(
                "Business Performance & Sales Analysis",
                style={
                    "textAlign": "center",
                    "color": "#777",
                    "fontSize": "16px"
                }
            )
        ]),

        html.Hr(),

        # Dashboard Filters
        html.Div([

            html.H3(
                "Dashboard Filters",
                style={"color": "#4b2815"}
            ),

            html.Div([

                # Coffee Filter
                html.Div([
                    html.Label("Select Coffee Type"),

                    dcc.Dropdown(
                        id="coffee_filter",
                        options=coffee_options,
                        value="All",
                        clearable=False
                    )
                ], style={"width": "32%"}),

                # Time Filter
                html.Div([
                    html.Label("Select Time of Day"),

                    dcc.Dropdown(
                        id="time_filter",
                        options=time_options,
                        value="All",
                        clearable=False
                    )
                ], style={"width": "32%"}),

                # Cash Type Filter
                html.Div([
                    html.Label("Select Cash Type"),

                    dcc.Dropdown(
                        id="cash_filter",
                        options=cash_options,
                        value="All",
                        clearable=False
                    )
                ], style={"width": "32%"})

            ], style={
                "display": "flex",
                "gap": "20px",
                "justifyContent": "space-between"
            })

        ], style={
            "backgroundColor": "white",
            "padding": "20px",
            "borderRadius": "12px",
            "marginBottom": "25px"
        }),

        # KPI Cards
        html.H3(
            "Business Performance Overview",
            style={"color": "#4b2815"}
        ),

        html.Div(
            id="kpi_cards",
            style={
                "display": "flex",
                "gap": "15px",
                "flexWrap": "wrap",
                "marginBottom": "25px"
            }
        ),

        # First Row Charts
        html.Div([

            html.Div([
                dcc.Graph(
                    id="daily_sales",
                    config={"displayModeBar": False}
                )
            ], style={
                "width": "60%",
                "backgroundColor": "white",
                "padding": "10px",
                "borderRadius": "12px"
            }),

            html.Div([
                dcc.Graph(
                    id="coffee_sales",
                    config={"displayModeBar": False}
                )
            ], style={
                "width": "40%",
                "backgroundColor": "white",
                "padding": "10px",
                "borderRadius": "12px"
            })

        ], style={
            "display": "flex",
            "gap": "15px",
            "marginBottom": "20px"
        }),

        # Second Row Charts
        html.Div([

            html.Div([
                dcc.Graph(
                    id="time_sales",
                    config={"displayModeBar": False}
                )
            ], style={
                "width": "50%",
                "backgroundColor": "white",
                "padding": "10px",
                "borderRadius": "12px"
            }),

            html.Div([
                dcc.Graph(
                    id="hourly_sales",
                    config={"displayModeBar": False}
                )
            ], style={
                "width": "50%",
                "backgroundColor": "white",
                "padding": "10px",
                "borderRadius": "12px"
            })

        ], style={
            "display": "flex",
            "gap": "15px",
            "marginBottom": "20px"
        }),

        # Footer
        html.Hr(),

        html.P(
            "Developed using Python, Dash, Pandas and Plotly",
            style={
                "textAlign": "center",
                "color": "#777",
                "fontSize": "13px"
            }
        )

    ]
)
```

**Step 6**: Intractive Dashboard

```python
# Dashboard Callback

@app.callback(
    [
        Output("kpi_cards", "children"),
        Output("daily_sales", "figure"),
        Output("coffee_sales", "figure"),
        Output("time_sales", "figure"),
        Output("hourly_sales", "figure")
    ],
    [
        Input("coffee_filter", "value"),
        Input("time_filter", "value"),
        Input("cash_filter", "value")
    ]
)

def update_dashboard(coffee, period, cash):

    # Apply Filters
    filtered = df.copy()

    if coffee != "All":
        filtered = filtered[
            filtered["coffee_name"] == coffee
        ]

    if period != "All":
        filtered = filtered[
            filtered["Time_of_Day"] == period
        ]

    if cash != "All":
        filtered = filtered[
            filtered["cash_type"] == cash
        ]

    # Calculate KPIs
    total_sales = filtered["money"].sum()

    transactions = len(filtered)

    avg_sales = (
        total_sales / transactions
        if transactions > 0 else 0
    )

    top_coffee = (
        filtered["coffee_name"].value_counts().idxmax()
        if transactions > 0 else "No Data"
    )

    # KPI Cards
    def make_card(title, value):
        return html.Div([
            html.H4(title, style={"color": "#666"}),
            html.H2(value, style={"color": "#5c3317"})
        ], style={
            "backgroundColor": "white",
            "padding": "18px",
            "textAlign": "center",
            "borderRadius": "12px",
            "flex": "1",
            "minWidth": "180px",
            "boxShadow": "0 2px 5px #ccc"
        })

    cards = [
        make_card("Total Revenue", f"{total_sales:,.2f}"),
        make_card("Total Transactions", f"{transactions:,}"),
        make_card("Average Order Value", f"{avg_sales:,.2f}"),
        make_card("Best Selling Coffee", top_coffee)
    ]

    # Daily Sales Chart
    daily = filtered.groupby(
        "Date", as_index=False
    )["money"].sum()

    fig1 = px.line(
        daily,
        x="Date",
        y="money",
        title="Daily Coffee Sales",
        markers=True,
        labels={
            "money": "Sales Revenue",
            "Date": "Date"
        }
    )

    # Coffee-wise Sales Chart
    coffee_data = filtered.groupby(
        "coffee_name", as_index=False
    ).agg(
        Revenue=("money", "sum"),
        Transactions=("money", "count")
    ).sort_values("Revenue", ascending=False)

    fig2 = px.bar(
        coffee_data,
        x="coffee_name",
        y="Revenue",
        title="Revenue by Coffee Type",
        color="coffee_name",
        labels={
            "coffee_name": "Coffee Type",
            "Revenue": "Total Revenue"
        }
    )

    fig2.update_layout(showlegend=False)

    # Time of Day Sales Chart
    time_data = filtered.groupby(
        "Time_of_Day", as_index=False
    )["money"].sum()

    fig3 = px.pie(
        time_data,
        names="Time_of_Day",
        values="money",
        title="Sales by Time of Day",
        hole=0.4
    )

    # Hourly Sales Chart
    hourly = filtered.groupby(
        "hour_of_day", as_index=False
    )["money"].sum()

    fig4 = px.bar(
        hourly,
        x="hour_of_day",
        y="money",
        title="Hourly Sales Analysis",
        labels={
            "hour_of_day": "Hour",
            "money": "Sales Revenue"
        }
    )

    # Chart Styling
    for fig in [fig1, fig2, fig3, fig4]:
        fig.update_layout(
            template="plotly_white",
            margin=dict(l=30, r=30, t=60, b=30)
        )

    # Handle Empty Data
    if filtered.empty:
        for fig in [fig1, fig2, fig3, fig4]:
            fig.add_annotation(
                text="No data available",
                showarrow=False
            )

    return cards, fig1, fig2, fig3, fig4


# Run Dashboard
if __name__ == "__main__":
    app.run(debug=False)
```
