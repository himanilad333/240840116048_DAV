step 1 - import pandas

```python
# Import the Pandas library
import pandas as pd
```

Step - 2 Create sample dataset

```python
# Create a dictionary containing student data
data = {
    "ID":[101,102,103,104,105,106,107],
    "Name":["Rahul","Priya","Amit","Riya","Karan","Krisha","Harry"],
    "Age":[20,21,19,20,22,24,28],
    "City":["Surat","Navsari","Valsad","Bharuch","Surat","Gandhinagar","Ahemdabad"],
    "Marks":[85,92,78,88,81,84,90]
}
```

step-3 create a dataFrame

```python
# Create a DataFrame from the dictionary 'data'
df = pd.DataFrame(data)
# Display the DataFrame
df
```

step-4 save data as a csv file

```python
# Save the DataFrame to a CSV file without row indices
df.to_csv("student.csv", index=False)
# Display a success message
print("CSV file created successfully")
```

step-5 save data as an excel file

```python
# Save the DataFrame to an Excel file without row indices
df.to_excel("student.xlsx", index=False)

# Display a success message
print("Excel file created successfully")
```

step-6 save data as a JSON file

```python
# Save the DataFrame to a JSON file in records format with indentation
df.to_json("student.json", orient="records", indent=4)
# Display a success message
print("JSON file created successfully")
```

step-7 Read the CSV file

```python
# Read data from the CSV file into a DataFrame
df_csv = pd.read_csv("student.csv")

# Display the DataFrame
df_csv
```
